## External dataset

The added external dataset is stored as:

```text
data/external_student_performance_uci.csv
```

It combines the UCI Mathematics and Portuguese student-performance files into one course-level dataset with a `subject` column. The original `G3` final grade is transformed into `performance_level`:

```text
G3 < 10        -> Low
10 <= G3 < 15  -> Medium
G3 >= 15       -> High
```

## Interpretation to use in the report

The old 8-class prediction problem was too sparse because the original dataset has only 145 rows. Grouping grades into Low, Medium, and High makes the task more stable.

The external UCI dataset gives a stronger comparative ML analysis because it has 1044 course-level observations. The best external setting includes `G1` and `G2`, which are previous period grades and therefore strong predictors of `G3`. This is why the external model reaches around 0.89 test accuracy. The notebook also reports a harder early-prediction setting without `G1` and `G2`; that score is lower but more realistic when previous grades are not available.
